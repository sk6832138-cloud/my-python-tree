practice Pandas

```python
import pandas as pd

data = {
    "Name": ["Shubham", "Priya", "Rahul"],
    "Marks": [85, 92, 78]
}

df = pd.DataFrame(data)
print(df)
print(df["Marks"].mean()) # average
```

Practice data cleaning

```python
data = {
    "Name": ["Shubham", "Priya", "Rahul"],
    "Marks": [85, 92, 78]
}

df = pd.DataFrame(data)
print("Before Cleaning:")
print(df)

df=df.dropna() # Remove empty rows
print("After Cleaning:")
print(df)
```

Practice graph- Matplotlib

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {"Name": ["Shubham", "Priya", "Rahul"],
        "Marks": [85, 92, 78]}

df = pd.DataFrame(data)

plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks")
plt.show()
```
